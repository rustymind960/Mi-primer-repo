```python
#Ejercicio 1 — Limpieza de nombres de clientes

#El banco recibió esta lista de nombres de clientes con errores de digitación (espacios y mayúsculas inconsistentes):
#clientes = [" MARIA lopez", "juan PEREZ ", "Andrea Gómez ", " CARLOS ruiz"]
#Recorre la lista y para cada nombre, y normaliza los nombres. Guarda los resultados en una nueva lista clientes_limpios e imprímela.

clientes=[" MARIA lopez", "juan PEREZ ", "Andrea Gómez ", " CARLOS ruiz"]
clientes_limpios=[]

for cliente in clientes:
  cliente_limpio = cliente.title().strip()
  clientes_limpios.append(cliente_limpio)

print(clientes_limpios)
```

```python
from os import replace
#Ejercicio 2 — Normalización de números de cuenta

#Las cuentas llegaron con formatos distintos (guiones, espacios):
#cuentas = ["123-456-789", "987 654 321", "111-222-333 ", " 444555666"]
#Recorre la lista y quitar '-' y ' ', además quita los espacios. Guarda el resultado en cuentas_limpias e imprímela.

cuentas = ["123-456-789", "987 654 321", "111-222-333 ", " 444555666"]
cuentas_limpias = []

for cuenta in cuentas:
  cuenta_limpia = cuenta.replace("-","").replace(" ","")
  cuentas_limpias.append(cuenta_limpia)

print(cuentas_limpias)
```

```python
#Ejercicio 3 — Corrección de montos de transacciones

#Los montos llegaron como texto, con símbolo de pesos y separadores de miles:
#montos_texto = ["$1.250.000", "$980.000", "$2.100.000", "$500.000"]
#Recorre la lista y quita el signo de '$' y '.', luego convierte cada valor a int. Guarda los montos numéricos en
#montos_limpios, suma el total con for y usa if/else para indicar si el total supera un límite diario de $4.000.000.

montos_texto = ["$1.250.000", "$980.000", "$2.100.000", "$500.000"]
montos_limpios = []

for monto in montos_texto:
  monto_limpio = monto.replace("$","").replace(".","")
  montos_limpios.append(int(monto_limpio))

for monto_suma in montos_limpios:
  montos_suma = sum(montos_limpios)

if montos_suma>4000000:
    print("Supera el límite diario")

else:

    print("No supera el límite diario")
```

```python
#Ejercicio 4 — Eliminación de cédulas duplicadas

#Un reporte de clientes llegó con cédulas repetidas:
#cedulas_clientes = ["1020304050", "1030405060", "1020304050", "1050607080", "1030405060"]
#Recorre la lista con for y construye una lista cedulas_unicas sin duplicados. Al final imprime cuántas cédulas duplicadas
#se eliminaron.

cedulas_clientes = ["1020304050", "1030405060", "1020304050", "1050607080", "1030405060"]
cedulas_unicas = []
cedulas_duplicadas = []

for cedula in cedulas_clientes:
  if cedula not in cedulas_unicas:
    cedulas_unicas.append(cedula)
  else:
    cedulas_duplicadas.append(cedula)

cedula_duplicada = len(cedulas_clientes) - len(cedulas_unicas)


print(f"Cédulas únicas: {cedulas_unicas}")
print(f"Cédulas duplicadas: {cedulas_duplicadas}")

```

```python
#Ejercicio 5 — Filtrado de transacciones inválidas

#El sistema registró transacciones con valores inválidos (negativos o en cero):
#transacciones = [150000, -20000, 300000, 0, 450000, -5000]
#Recorre una copia de la lista con for e identifica los valores <= 0. Elimina los valores incorrectos, imprimiendo
#cuál valor se elimina en cada caso. Al final calcula e imprime el total de las transacciones válidas que quedaron.

transacciones = [150000, -20000, 300000, 0, 450000, -5000]

for transaccion in transacciones.copy():
  if transaccion<=0:
    print(f"Valor inválido eliminado: {transaccion}")
    transacciones.remove(transaccion)

transaccion_valida = len(transacciones)
total_transacciones = sum(transacciones)
print(f"Cantidad de transacciones válidas: {transaccion_valida}")
print(f"Valor total de las transacciones válidas: {total_transacciones}")

```

```python
#Ejercicio 6 — Clasificación de clientes por saldo

#El banco quiere segmentar clientes según su saldo actual:
#saldos = [8500000, 1200000, 300000, 15000000, 750000, 50000]
#Recorre la lista con for y usa if/elif/else para clasificar cada saldo:
#• saldo >= 10.000.000: "Premium"
#• 1.000.000 <= saldo < 10.000.000: "Preferencial"
#• saldo < 1.000.000: "Básico"
#Lleva un conteo de clientes por categoría e imprime el resumen final.

saldos = [8500000, 1200000, 300000, 15000000, 750000, 50000]
cliente_premium = 0
cliente_preferencial = 0
cliente_basico = 0

for saldo in saldos:
  if saldo>= 10000000:
    cliente_premium += 1
  elif 1000000 <= saldo < 10000000:
    cliente_preferencial += 1
  else:
    cliente_basico += 1

print(f"Cantidad de clientes con categoría premium: {cliente_premium}")
print(f"Cantidad de clientes con categoría preferencial: {cliente_preferencial}")
print(f"Cantidad de clientes con categoría básico: {cliente_basico}")


```

```python
#Ejercicio 7 — Estandarización de correos electrónicos

#Los correos de los clientes llegaron sin un formato uniforme:
#correos = [" Maria.Lopez@Banco.com", "JUAN.PEREZ@BANCO.COM ", "andrea_gomez@banco.COM", " carlos.Ruiz@Banco.Com "]
#Recorre la lista con for y usa strip() y lower() para dejar todos los correos en minúscula y sin espacios. Guarda el
#resultado en correos_limpios e imprímela.

correos = [" Maria.Lopez@Banco.com", "JUAN.PEREZ@BANCO.COM ", "andrea_gomez@banco.COM", " carlos.Ruiz@Banco.Com "]
correos_limpios =[]

for correo in correos:
  correo_limpio = correo.strip().lower()
  correos_limpios.append(correo_limpio)

print(correos_limpios)
```

```python
#Ejercicio 8 — Detección de teléfonos inválidos

#La base de datos de contacto tiene valores vacíos o inválidos:
#telefonos = ["3101234567", "", "3209876543", " ", "3157894561", "N/A"]
#Recorre la lista con for y usa if/elif/else para revisar cada valor: si al aplicar strip() queda vacío, o si es igual a 'N/A',
#reemplázalo por 'Sin registrar'; de lo contrario consérvalo tal cual. Guarda los resultados en telefonos_validados usando
#append() e imprímela.

telefonos = ["3101234567", "", "3209876543", " ", "3157894561", "N/A"]
telefonos_validados=[]

for telefono in telefonos:
  telefono_limpio=telefono.strip()
  if telefono_limpio=="":
    telefonos_validados.append("Sin registrar")
  elif telefono_limpio=="N/A":
    telefonos_validados.append("Sin registrar")
  else:
    telefonos_validados.append(telefono)

print(telefonos_validados)
```

```python
#Ejercicio 9 — Verificación del estado de las cuentas

#El estado de las cuentas quedó registrado sin un formato consistente:
#estados = ["activo", "ACTIVO", "Inactivo ", " inactivo", "Activo", "INACTIVO"]
#Recorre la lista con for, normaliza cada valor con strip() y lower(), y usa if/elif/else para contar cuántas cuentas están
#activas y cuántas inactivas. Imprime el resumen final.

estados = ["activo", "ACTIVO", "Inactivo ", " inactivo", "Activo", "INACTIVO"]
activo = 0
inactivo = 0

for estado in estados:
  estado_limpio = estado.strip().lower()
  if estado_limpio == "activo":
    activo += 1
  elif estado_limpio == "inactivo":
    inactivo += 1

print(f"Cuentas activas: {activo}")
print(f"Cuentas inactivas: {inactivo}")
```

```python
#Ejercicio 10 — Consolidación de lista de sucursales

#El área de operaciones envió esta lista de sucursales con nombres repetidos y mal formateados:
#sucursales = ["Chapinero", "chapinero", "Suba", "SUBA", "Kennedy", "Chapinero "]
#Recorre la lista con for y usa strip() y title() para normalizar cada nombre. Usa if ... not in ... y append() para construir
#una lista sucursales_unicas sin duplicados. Finalmente, usa sorted() para ordenarla alfabéticamente e imprímela.

sucursales = ["Chapinero", "chapinero", "Suba", "SUBA", "Kennedy", "Chapinero "]
sucursales_unicas = []

for sucursal in sucursales:
  sucursal_limpia=sucursal.strip().title()
  if sucursal_limpia not in sucursales_unicas:
    sucursales_unicas.append(sucursal_limpia)

print(sorted(sucursales_unicas))
```

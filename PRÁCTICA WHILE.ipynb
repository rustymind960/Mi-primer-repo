Una tienda de barrio, tiene el inventario de 6 productos en una lista: inventario = [12,3,8,0,25,4]. Recorre la lista e imprime por cada producto si está agotado (que hayan 0 unidades), si está en nivel crítico (menos de 5 unidades), o si tiene inventario suficiente. Al final, di cuántos productos hay que pedir (los que están agotados o los que están en estado crítico)

```python
inventario = [12,3,8,0,25,4]
posicion=0
por_pedir=0

while posicion < len(inventario):
  producto=inventario[posicion] #Sacar cada elemento de la lista

  if producto == 0:
    print(producto, "Producto agotado")
    por_pedir+=1
  elif producto<5:
    print(producto, "Producto en estado crítico")
    por_pedir+=1
  else:
    print(producto, "Producto con suficiente inventario")

  posicion+=1

print(f"Cantidad de productos por comprar: {por_pedir}")

```

Tienes las temperaturas de 7 días. Clasifica cada una como "Fría", cuando es menos de 12, "Templada" cuando está entre 12 y 16 o "Cálida", cuando es más de 16, y cuenta cuántos días hubo de cada tipo.

```python
temperaturas = [14,12,18,9,15,11,17]
posicion = 0
dia_frio = 0
dia_calido= 0
dia_templado= 0

while posicion < len(temperaturas):
  temperatura = temperaturas[posicion]

  if temperatura < 12:
    dia_frio+=1
    print("Temperatura fría:", temperatura)
  elif 12<=temperatura<=16:
    dia_templado+=1
    print("Temperatura templada:", temperatura)
  else:
    dia_calido+=1
    print("Temperatura cálida:", temperatura)

  posicion+=1

print("Cantidad de días fríos:", dia_frio)
print("Cantidad de días templados:", dia_templado)
print("Cantidad de días cálidos:", dia_calido)
```

Tienes los pasajeros que entraron a una estación en 6 franjas horarias. Encuentra la franja con más pasajeros y cuenta cuántas franjas superaron los 600.

```python
pasajeros=[320,480,910,650,1200,540]

posicion = 0
num_pasajero = 0
franja_maxima = 0
mas_600 = 0

while posicion <  len(pasajeros):
  pasajero = pasajeros[posicion]

  if pasajero>num_pasajero:
    franja_maxima = posicion+1
    num_pasajero=pasajero

  if pasajero>600:
    mas_600+=1

  posicion+=1

print("La franja con más pasajeros fue la franja número", franja_maxima, "con", num_pasajero, "pasajeros")
print("El número de franjas que superaron 600 pasajeros fue:", mas_600)
```

Tienes registrado el número de pasos que realizaste cada día de la semana, de lunes a domingo. Recorre la lista utilizando obligatoriamente un ciclo while y muestra, día por día, si el nivel de actividad física fue:

Bajo: menos de 5.000 pasos.

Moderado: entre 5.000 y 10.000 pasos.

Alto: más de 10.000 pasos.

Requerimientos del programa:Mostrar día a día el número de pasos y su clasificación (Bajo, Moderado o Alto) usando un ciclo while.   Calcular y mostrar el total de pasos caminados en la semana.   Contar y mostrar cuántos días tuviste una actividad física alta.   Indicar qué día realizaste la mayor cantidad de pasos y cuántos pasos fueron.

```python
pasos = [4200, 8500, 12300, 3100, 9800, 15400, 6700]
act_alta=0
posicion = 0
num_pasos_max=0
dia_pasos_max=pasos[0]

while posicion<len(pasos):
  paso=pasos[posicion]

  if paso<5000:
    print("Nivel de actividad física baja:", paso, "pasos")
  elif 5000<=paso<=10000:
    print("Nivel de actividad física moderado:", paso, "pasos")
  else:
    act_alta+=1
    print("Nivel de actividad física alto:", paso, "pasos")

  posicion+=1

  if paso>num_pasos_max:
    num_pasos_max=paso
    dia_pasos_max=posicion


num_pasos= sum(pasos)
print("Cantidad de pasos dados en la semana:", num_pasos)
print("La cantidad de días donde se tuvo actividad física alta fueron:", act_alta)
print("El día con la mayor cantidad de pasos fue el", dia_pasos_max, "con", num_pasos_max, "pasos")
```

Una estación meteorológica registra la temperatura máxima (°C) alcanzada cada día de la semana, de lunes a domingo. Recorre las listas usando un ciclo while y clasifica el día según la temperatura:

Frío: menos de 15 °C.

Templado: entre 15 °C y 25 °C.

Cálido: más de 25 °C.

Requerimientos del programa:
Recorrer las listas mediante un ciclo while e imprimir el día, la temperatura y su clasificación (Frío, Templado o Cálido).

Calcular y mostrar la temperatura promedio de la semana.

Contar cuántos días fueron cálidos.

Indicar cuál fue el día más caluroso (máxima temperatura) y de cuánto fue.

Indicar cuál fue el día más frío (mínima temperatura) y de cuánto fue.

```python
temperaturas = [14, 22, 28, 12, 19, 26, 18]
posicion=0
calidos=0
dia_max_temp=0
max_temp=0
dia_min_temp=0
min_temp=temperaturas[0]

while posicion<len(temperaturas):
  temperatura=temperaturas[posicion]

  if temperatura<15:
    print("Temperatura fría:", temperatura,"°C")
  elif 15>=temperatura>=25:
    print("Temperatura templada:", temperatura,"°C")
  elif temperatura>25:
    calidos+=1
    print("Temperatura cálida:", temperatura,"°C")

  posicion+=1

  if temperatura>max_temp:
    max_temp=temperatura
    dia_max_temp=posicion
  if temperatura<min_temp:
    min_temp=temperatura
    dia_min_temp=posicion

prom_temp=sum(temperaturas)/len(temperaturas)

print("\nEl promedio de temperatura de la semana fue de", round(prom_temp, 1),"°C")
print("La cantidad de días cálidos fueron", calidos)
print("El día más caluroso fue el día", dia_max_temp, "con una temperatura de", max_temp,"°C")
print("El día más frío fue el día", dia_min_temp, "con una temperatura de", min_temp,"°C")
```

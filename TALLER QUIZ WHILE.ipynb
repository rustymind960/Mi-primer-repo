Caso 1. Los gastos de tu semana
Tienes registrado lo que gastaste cada día de la semana, de lunes a domingo. Recorre la lista con un ciclo while y muestra, día por día, si el gasto fue bajo (menos de 30.000), moderado (entre 30.000 y 60.000) o alto (más de 60.000).
Datos:  gastos = [32000, 58000, 15000, 72000, 41000, 95000, 28000]
El programa debe mostrar:
•	Cuánto gastaste en total durante la semana.
•	Cuántos días tuviste gasto alto.
•	Cuál fue el día de mayor gasto y de cuánto fue.


```python
gastos = [32000, 58000, 15000, 72000, 41000, 95000, 28000]

posicion=0
alto=0
dia_max=0
gasto_max=0

while posicion < len(gastos):
  gasto = gastos[posicion]


  if gasto < 30000:
    print("Gasto bajo:", gasto)
  elif 30000<=gasto<=60000:
    print("Gasto moderado:", gasto)
  else:
    alto+=1
    dia_max=posicion+1
    gasto_max=gasto


    print("Gasto alto:", gasto)

  posicion+=1

print(f"El gasto total durante la semana fue de {sum(gastos)}")
print("El número de días con gastos altos fueron", alto)
print("El día con mayor gasto fue el día", dia_max, "con un gasto de", gasto_max)
```

Caso 2. El mercado del mes
Guardaste el precio de ocho productos de tu mercado. Recorre la lista con un ciclo while e identifica en qué se te va la plata.
Datos:  precios = [4500, 12000, 8900, 23000, 3200, 15500, 6700, 31000]
El programa debe mostrar:
•	El valor total del mercado.
•	El producto más barato y el más caro, con su precio y su posición en la lista.
•	Cuántos productos costaron más de 10.000 pesos.


```python
precios = [4500, 12000, 8900, 23000, 3200, 15500, 6700, 31000]

posicion=0
mas_barato=0
precio_barato=precios[0]
mas_caro=0
precio_caro=0
mas_10000=0

while posicion<len(precios):
  producto=precios[posicion]

  if producto<precio_barato:
    mas_barato=posicion+1
    precio_barato=producto
  if producto>precio_caro:
    mas_caro=posicion+1
    precio_caro=producto
  if producto>10000:
    mas_10000+=1

  posicion+=1

total_mercado=sum(precios)
print("El valor total del mercado es de", total_mercado)
print("El producto más barato fue el", mas_barato, "con un precio de", precio_barato)
print("El producto más caro fue el", mas_caro, "con un precio de", precio_caro)
print(mas_10000, "fueron la cantidad de productos que costaron más de 10000")
```

Caso 3. La meta de ahorro
Te propusiste ahorrar 3.000.000 de pesos. Cada mes el programa te pregunta cuánto lograste ahorrar ese mes y lo va sumando. El ciclo termina cuando alcances o superes la meta.
Datos:  meta = 3000000
El programa debe mostrar:
•	Cuántos meses te tomó llegar a la meta.


```python
meta = 3000000
ahorro=0
meses=0

while ahorro<meta:
  ahorro+=int(input("Por favor ingrese su ahorro de este mes: "))
  meses+=1

print("Cantidad de meses en conseguir la meta de ahorro:", meses)
```

Caso 4. Control de gastos diarios
El programa te pide gastos uno por uno y los va registrando. Termina cuando escribas 0. Por cada gasto debe decir si fue un gasto hormiga (menos de 10.000), un gasto normal (entre 10.000 y 50.000) o un gasto fuerte (más de 50.000).
Datos:  Centinela: escribir 0 para terminar
El programa debe mostrar:
•	Cuántos gastos registraste en total.
•	La suma de todos los gastos.
•	El gasto más alto y el gasto más bajo que registraste.


```python
gastos=int(input("Por favor escriba el valor del gasto realizado: "))
num_gastos=0
suma_gastos=0
gasto_menor=0
gasto_mayor=0

while gastos!=0:

  if gastos<10000:
    print("Gasto hormiga")
  elif 10000<=gastos<=50000:
    print("Gasto normal")
  else:
    print("Gasto fuerte")
  num_gastos+=1

  if num_gastos==1:
    gasto_mayor=gastos
    gasto_menor=gastos
  else:
    if gastos>gasto_mayor:
      gasto_mayor=gastos
    if gastos<gasto_menor:
      gasto_menor=gastos
  gastos=int(input("Por favor escriba el valor del gasto realizado: "))
  suma_gastos+=gastos


print("Número de gastos:", num_gastos)
print("Suma de gastos:", suma_gastos)
print("Gasto más alto:", gasto_mayor)
print("Gasto más bajo:", gasto_menor)
```

Debes 1.200.000 pesos en la tarjeta de crédito. Cada mes el banco te cobra 10.000 de interés sobre el saldo y tú abonas 200.000 pesos fijos. El ciclo repite mes a mes hasta que el saldo llegue a cero.
Datos:  saldo = 1200000  |  interes = 10000 mensual  |  abono = 200000
El programa debe mostrar:
•	Cuántos meses te tomará quedar libre de la deuda.
•	Cuánto pagaste en intereses en total (ese es el acumulador clave).
•	Imprime en cada mes el saldo que queda pendiente.


```python
saldo = 1200000
interes = 10000
abono = 200000

meses = 0
total_intereses = 0

print("--- SEGUIMIENTO MES A MES ---")

while saldo > 0:
    meses += 1

    saldo += interes
    total_intereses += interes

    saldo -= abono


    if saldo < 0:
        saldo = 0

    print(f"Mes {meses}: Saldo pendiente = ${saldo}")

print("\n--- RESUMEN FINAL ---")
print(f"Meses necesarios para saldar la deuda: {meses} meses")
print(f"Total pagado en intereses: ${total_intereses}")
```

Caso 6. Separar los gastos hormiga (nivel intermedio)
Tienes diez gastos del mes mezclados. Recórrelos con un ciclo while y sepáralos en dos listas nuevas, que empiezan vacías: una con los gastos hormiga (menos de 20.000) y otra con los gastos grandes (20.000 o más).
Datos:  gastos = [3500, 45000, 2800, 120000, 6200, 38000, 4100, 85000, 1500, 52000]
El programa debe mostrar:
•	Las dos listas nuevas, impresas al final.
•	Cuánto suma cada lista y cuántos gastos quedaron en cada una.


```python

gastos = [3500, 45000, 2800, 120000, 6200, 38000, 4100, 85000, 1500, 52000]

# Listas vacías para clasificar los gastos
gastos_hormiga = []
gastos_grandes = []

posicion = 0

while posicion < len(gastos):
    monto = gastos[posicion]

    if monto < 20000:
        gastos_hormiga.append(monto)
    else:
        gastos_grandes.append(monto)

    posicion += 1


total_hormiga = sum(gastos_hormiga)
total_grandes = sum(gastos_grandes)

cantidad_hormiga = len(gastos_hormiga)
cantidad_grandes = len(gastos_grandes)

print("--- CLASIFICACIÓN DE GASTOS ---")
print("Gastos hormiga (< $20.000):", gastos_hormiga)
print("Gastos grandes (>= $20.000):", gastos_grandes)

print("\n--- RESUMEN ---")
print(f"Gastos hormiga: {cantidad_hormiga} gastos | Suma total: ${total_hormiga}")
print(f"Gastos grandes: {cantidad_grandes} gastos | Suma total: ${total_grandes}")
```

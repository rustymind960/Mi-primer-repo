```python
def validar_nombre(nombre):
    if nombre.isalpha():
        return nombre
    else:
        return "Error"

validar_nombre("Pedro1")
```

```python
nombre=input("Ingrese un nombre:")
validar_nombre(nombre)
```

```python
nombres=["Juan", "Ana", "Enrique 8", "Maria", "Luis4"]

def validar_nombres(nombres):
    for nombre in nombres:
        validar_nombre(nombre)

validar_nombres(nombres)
```

```python
def validar_nombre_while():
    nombre=input("Ingrese su nombre:")
    intentos=1

    while (not nombre == validar_nombre(nombre)) and intentos<3: #Nota: Una forma equivalente y más común de escribir esa misma línea sin usar not es usando el operador "diferente de": while nombre != validar_nombre(nombre):
        nombre=input("Ingrese un nombre:")
        intentos+=1

    print(intentos)


```

```python
#Ejercicio 1

correo = input("Por favor ingrese su correo: ")

correo_limpio = correo.strip().lower()

if correo_limpio.endswith("@universidad.edu.co"):
  print("Correo institucional válido")
else:
  print("Correo no válido para acceso institucional")
```

```python
#Ejercicio 2

tipo_usuario = input("Tipo de usuario: ")
estado = input("Estado activo o inactivo: ")

usuario_limpio = tipo_usuario.lower()
estado_limpio = estado.lower()

if usuario_limpio == "estudiante" and estado_limpio == "activo":
  print("Acceso concebido a la biblioteca digital")
else:
  print("Acceso denegado")
```

```python
#Ejercicio 3

promedio = float(input("Ingrese su promedio: "))
estrato = int(input("Ingrese su estrato: "))

if promedio>=4.5 and estrato<=2:
  print("Felicidades, tienes beca académica")
else:
  print("No aplica para beca")
```

```python
#Ejercicio 4

codigo = str(input("Ingrse su código: "))

if len(codigo) == 8 and codigo.isnumeric() == True:
  print("Código válido")
else:
  print("Código inválido")
```

```python
#Ejercicio 5

curso = str(input("Ingrese el nombre del curso: "))
nota = float(input("Ingrese su nota: "))

curso_limpio = curso.strip().lower().replace(" ","").replace("ó","o")

if curso_limpio == "programacion1" and nota>3.5:
  print("Puede inscribirse al curso avanzado")
else:
  print("No cumple requisitos")
```

```python
#Ejercicio 6

dia_pago = int(input("Ingrese día de pago: "))
promedio = float(input("Ingrese su promedio: "))

if dia_pago<10 and promedio>4.3:
  print("Aplica descuento en matrícula")
else:
  print("No aplica descuento")
```

```python
#Ejercicio 7

nombre = input("Ingrese su nombre: ")

if nombre.isalpha() == True and len(nombre)>3:
  print("Nombre válido")
else:
  print("Nombre inválido")
```

```python
#Ejercicio 8

edad = int(input("Ingrese su edad: "))

if 16<=edad<=25:
  print("Estudiante traidicional")
elif edad>25:
  print("Estudiante adulto")
```

```python
#Ejercicio 9

salon = input("Ingrese su salón: ")
salon_limpio = salon.title()

if salon_limpio.startswith("A") or salon_limpio.startswith("B"):
  print("Salón válido para el curso")
else:
  print("Salón no permitido para este curso")
```

```python
#Ejercicio 10

bata = input("¿Posee bata? ")
carnet = input("¿Posee carnet? ")

bata_limpia = bata.lower().replace("í","i")
carnet_limpio = carnet.lower().replace("í","i")

if bata_limpia == "si" and carnet_limpio == "si":
  print("Acceso permitido al laboratorio")
elif bata_limpia =="no" and carnet_limpio =="no":
  print("Acceso denegado: faltan requisitos de seguridad")
else:
  print("Acceso denegado: faltan requisitos de seguridad")
```

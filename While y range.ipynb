Pide la edad del usuario. Si escribe un número fuera del rango (del 1 al 100), vuelve a pedir la edad cuando sea válida, imprime "Edad registrada" y el valor

```python
edad = int(input("Ingrese su edad: "))

while edad < 1 or edad > 100:
  print("Ingrese un valor válido")
  edad = int(input("Ingrese su edad: "))

print("Edad registrada")
```

Pide números al usuario y súmalos. El ciclo termina cuando escriba 0. Al final, muestra la suma y cuántos números ingresó, sin contar el 0

```python
suma=0
cantidad = 0
numero=int(input("Ingrese un número (termina cuando ingresa 0): "))

while numero !=0:
  suma += cantidad
  cantidad+=1
  numero=int(input("Ingrese un número (termina cuando ingresa 0): "))

print(f"Suma total: {suma}")
print(f"Cantidad: {cantidad}")
```

```python
canciones = ["Bad Bunny", "Karol G", "bad bunny", "Shakira", "KAROL G"]

bad_bunny = 0
karol_g = 0

for cancion in canciones:
    cancion_limpia = cancion.lower()
    if cancion_limpia == "bad bunny":
     bad_bunny += 1
    elif cancion_limpia == "karol g":
     karol_g += 1

print(f"Bad Bunny = {bad_bunny}")
print(f"Karol G = {karol_g}")
```

```python
repeticiones = [12, 15 ,8, 20, 5]

for repeticion in repeticiones:
    if repeticion >= 15:
        print("Serie fuerte:", repeticion)
    elif 10<=repeticion<14:
        print("Serie media:", repeticion)
    else:
        print("Seria floja:", repeticion)

repeticiones.sort(reverse=True)
print(f"Repeticiones ordenadas = {repeticiones}")
```

```python
nombres_usuario = ["ana123", "LUIS_99", "pedro", "MARTA2024", "juan"]
usuarios_validos = []

for usuario in nombres_usuario:
  usuario_limpio=usuario.lower()
  if len(usuario_limpio)>5:
    usuarios_validos.append(usuario_limpio)
  else:
    print("Nombre muy corto")

print(usuarios_validos)
```

```python
puntajes = [1200, 3400, 800, 5000, 2100]

for puntaje in puntajes:
  if puntaje<1000:
    print("Bronce")
  elif 1000<=puntaje<=2999:
    print("Plata")
  elif 3000<=puntaje<=4999:
    print("Oro")
  else:
    print("Diamante")

puntaje_maximo=max(puntajes)
maximo_posicion=puntajes.index(max(puntajes))+1

print(f"Puntaje máximo: {puntaje_maximo}")
print(f"Posición puntaje máximo: {maximo_posicion}")
```

```python
#6. Playlist de música — duración total
#Se tiene una lista con la duración en segundos de varias canciones de una playlist: 180, 240, 95,
#300 y 150. Recorre la lista y clasifica cada canción: si dura menos de 120 segundos imprime
#"Canción corta"; si dura entre 120 y 240 segundos imprime "Duración estándar"; si dura más de
#240 segundos imprime "Canción larga". Mientras recorres, ve sumando la duración de todas las
#canciones en un acumulador. Al final, convierte el total a minutos dividiendo entre 60 e imprime
#el resultado.

duraciones_canciones = [180, 240, 95, 300, 150]
duracion_total = 0

for duracion in duraciones_canciones:
  if duracion<120:
    print("Canción corta")
  elif 120<=duracion<=240:
    print("Duración estándar")
  else:
    print("Canción larga")

  duracion_total+=duracion

duracion_minutos = duracion_total/60
print(f"Duración total en minutos: {duracion_minutos}")
```

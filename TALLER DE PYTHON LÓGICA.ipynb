```python
#Ejercicio 1 — ¿Cumple la edad mínima para matricularse?

#Pide (o define en una variable) la edad de un aspirante como número entero. Imprimir:
#•	Si edad < 16: "No cumple la edad mínima para ingresar a la universidad."
#•	Si 16 <= edad <= 17: "Puede matricularse, pero necesita autorización de un acudiente."
#•	Si edad > 17: "Cumple la edad mínima para matricularse."

edad=21

if (edad<16):
  print("No cumple la edad mínima para ingresar a la universidad")

elif(edad>=16) and (edad<=17):
  print("Puede matricularse, pero necesita autorización de un acudiente")

else:
  print("Cumple la edad mínima para matricularse")

```

```python
#Ejercicio 2 — Lista de estudiantes del curso

#Crea una lista de strings llamada estudiantes con al menos 5 nombres. Recorre la lista e imprime cada nombre en mayúsculas. Al final, imprime cuántos estudiantes hay.

estudiantes=["Manuel", "Tatiana", "Oscar", "Tatiana", "Daniel"]

for estudiantes_mayus in estudiantes:
  print(estudiantes_mayus.upper())

print(f"Cantidad de estudiantes: {len(estudiantes)}")
```

```python
#Ejercicio 3 — Promedio del semestre

#Crea una lista de enteros llamada notas con las notas (de 0 a 5) de un estudiante en sus materias del semestre.
#Recorre la lista para sumar todas las notas, calcula el promedio e imprime si el estudiante aprobó o reprobó el semestre según un umbral (por ejemplo, promedio >= 3.0).

notas=[0, 1, 2, 3, 4, 5]

print(f"Promedio del semestre: {sum(notas)/len(notas)}")

notas_prom=sum(notas)/len(notas)

if notas_prom>=3:
  print("Aprobado")
else:
  print("Reprobado")
```

```python
#Ejercicio 6 — Clasificación de notas por rango

#Crea una lista de enteros llamada notas_parcial con al menos 8 notas (escala 0 a 100). Recorre la lista y usa if/elif/else para clasificar cada nota así:
#•	90-100: "A"
#•	70-89: "B"
#•	50-69: "C"
#•	0-49: "D"
#Lleva un conteo de cuántas notas caen en cada categoría (puedes usar variables contador ) e imprime el resumen final.

notas_parcial=[50, 67, 83, 78, 93, 95, 34, 99]

for nota_letra in notas_parcial:

  if(90<=nota_letra<=100):
    print("A")
  elif(70<=nota_letra<=89):
    print("B")
  elif(50<=nota_letra<=69):
    print("C")
  else:
    print("D")


```

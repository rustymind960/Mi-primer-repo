Punto 1 — Tabla de multiplicar
Pide un número al usuario e imprime su tabla de multiplicar del 1 al 10, con el formato 7 x 3 = 21.


```python
numero = int(input("Ingrese un número para ver su tabla de multiplicar: "))

for multiplo in range(1,11):
  resultado = numero * multiplo
  print(f"{multiplo} x {numero} = {resultado}")
```

Punto 2 — Cajero
Un cajero inicia con un saldo de 500.000. En cada vuelta pregunta cuánto quiere retirar el usuario y descuenta ese valor del saldo. El programa termina cuando el saldo llegue a 0 o quede por debajo del próximo retiro. Si el retiro es mayor al saldo disponible, muestra "Fondos insuficientes" y vuelve a preguntar.


```python
saldo = 500000

while saldo > 0:
  retiro = int(input("Ingrese cuánto quiere retirar: "))

  if retiro > saldo:
    print("Fondos insuficientes")
  else:
    saldo -= retiro
    print(f"Saldo después del retiro: {saldo}")

```

Punto 3 — Promedio del curso
Tienes esta lista de notas:
notas = [3.5, 4.2, 2.8, 4.9, 3.1, 2.5, 4.0]
Calcula e imprime el promedio, la nota más alta y cuántos estudiantes reprobaron (nota menor a 3.0). No uses sum(), max() ni len().


```python
notas = [3.5, 4.2, 2.8, 4.9, 3.1, 2.5, 4.0]
suma_notas= 0
cantidad_notas = 0

for nota in notas:
  suma_notas += nota
  cantidad_notas +=1
print(f"Promedio: {suma_notas/cantidad_notas}")

nota_maxima = notas[0]
for nota in notas:
  if nota > nota_maxima:
    nota_maxima = nota
print(f"Nota máxima: {nota_maxima}")

nota_minima = notas[0]
for nota in notas:
  if nota < nota_minima:
    nota_minima = nota

print(f"Nota mínima: {nota_minima}")
```

Punto 4 — Menú
Muestra este menú y ejecútalo hasta que el usuario elija la opción 4:
1. Saludar
2. Ver la fecha de hoy
3. Decir un chiste
4. Salir
Si el usuario escribe una opción que no existe, muestra "Opción inválida" y vuelve a mostrar el menú.


```python
while range(1,5):
  print("1. Saludar")
  print("2. Ver la fecha de hoy")
  print("3. Decir un chiste")
  print("4. Salir")

  opcion = int(input("Por favor ingrese una opcion: "))

  if opcion == 1:
    print("¡Hola!")
  elif opcion == 2:
    print("23/09/2026")
  elif opcion == 3:
    print("¿Qué le dice un GIF a un JPEG?")
    print("¡Vamos hombre, anímate!")
  elif opcion == 4:
    break
  else:
    print("Opción inválida. Por favor ingrese una opción del 1-4")
```

Punto 5 — Contraseña segura
Pide una contraseña al usuario. Es válida solo si tiene 8 caracteres o más. Si no la cumple, vuelve a pedirla. Dále máximo 3 intentos: si falla los tres, imprime "Cuenta bloqueada".


```python
intentos = 0

while intentos < 3:
  contraseña = input("Por favor digite su contraseña: ")

  if len(contraseña) >= 8:
    print("Contraseña registrada con éxito")
    break
  else:
    print("Contraseña inválida, debe tener 8 caracteres o más")
    intentos += 1

if intentos == 3:
    print("Cuenta bloqueada")
```

Punto 6 — Cuenta atrás de Collatz
Pide un número entero positivo. Mientras el número sea distinto de 1, aplícale esta regla e imprime el resultado en cada paso:
•	si es par, divide entre 2
•	si es impar, multiplícalo por 3 y súmale 1
Al final, imprime cuántos pasos tomó llegar a 1. Pruébalo con 6, con 7 y con 27.


```python
cuenta = int(input("Por favor ingrese un número entero positivo: "))
pasos = 0

while cuenta != 1:
  if cuenta % 2 == 0:
        cuenta = cuenta // 2
  else:
        cuenta = cuenta * 3 + 1

  pasos += 1
  print(cuenta)

print(f"Tomó {pasos} pasos llegar a 1")
```

```python
#Recorre la lista, quita guiones y espacios, y usa if/elif/else para verificar si cada tarjeta tiene exactamente 16 dígitos.
#Guarda las válidas en tarjetas_validas y las inválidas en tarjetas_invalidas, e imprime ambas listas.

tarjetas = ["4532-1111-2222-3333", "4532111122223333", "4532 1111 2222", "4532-1111-2222-33"]
tarjetas_validas = []
tarjetas_invalidas = []

for tarjeta in tarjetas:
  tarjeta_limpia = tarjeta.strip().replace("-","").replace(" ","")

  if len(tarjeta_limpia)==16:
    tarjetas_validas.append(tarjeta_limpia)

  else:
    tarjetas_invalidas.append(tarjeta_limpia)

print(f"Tarjetas válidas: {tarjetas_validas}")
print(f"Tarjetas inválidas: {tarjetas_invalidas}")
```
